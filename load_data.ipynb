Breast Cancer Wisconsin (Diagnostic) — UCI Machine Learning Repository

- `diagnosis`: `M` = 악성, `B` = 양성
- 속성 설명은 저장소의 `README.md`를 참고하세요.

## 1. 데이터 불러오기
GitHub 저장소에서 CSV를 바로 불러옵니다.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/thiskorea81/Breast_cancer/main/breast_cancer_wisconsin.csv"
df = pd.read_csv(url)

print(df.shape)
df.head()
```

## 2. 기본 정보와 결측치 확인

```python
df.info()
print("결측치 개수 합계:", df.isnull().sum().sum())
```

## 3. 진단 결과(정답) 분포

```python
df["diagnosis"].value_counts()
```

## 4. 기술 통계

```python
df.describe().T
```

## 5. 특성(X)과 정답(y) 나누기

```python
X = df.drop(columns=["id", "diagnosis"])        # 30개 특성
y = df["diagnosis"].map({"M": 1, "B": 0})       # 악성=1, 양성=0

print("X 크기:", X.shape)
print("y 크기:", y.shape)
```

## 6. 악성/양성별 평균 특성 비교

```python
mean_cols = [c for c in df.columns if c.endswith("_mean")]
df.groupby("diagnosis")[mean_cols].mean().T
```
